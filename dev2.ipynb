```python
import numpy as np

# Define your signal
t = np.linspace(0, 1, 500)  # Time variable
f = np.sin(2 * np.pi * 50 * t)  # Example: sine wave with frequency 50 Hz

# Parameters
N = len(f)  # Number of samples
F_manual = np.zeros(N, dtype=complex)  # To store Fourier Transform results
omega = np.linspace(-np.pi, np.pi, N)  # Frequencies (angular)

# Manually compute Fourier Transform (discrete approximation)
for k in range(N):  # For each frequency component
    for n in range(N):  # For each time step
        F_manual[k] += f[n] * np.exp(-2j * np.pi * k * n / N)

# Get the magnitude of the Fourier Transform
magnitude_manual = np.abs(F_manual)

# Plot the manually computed Fourier Transform
import matplotlib.pyplot as plt
plt.plot(omega, magnitude_manual)
plt.xlabel('Frequency (rad/s)')
plt.ylabel('Magnitude')
plt.title('Manual Fourier Transform')
plt.show()
```
